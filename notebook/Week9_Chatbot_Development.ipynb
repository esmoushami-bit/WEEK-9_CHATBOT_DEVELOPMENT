```python

import pandas as pd
import os
from datetime import datetime

print("Libraries imported successfully!")
```

```python
questions = pd.read_csv("../dataset/sample_questions.csv")

display(questions)

print("Total sample questions:", len(questions))
```

```python
history = pd.read_csv("../dataset/chat_history.csv")

display(history)

print("Total conversations:", len(history))
```

```python

print("Total conversations:", len(history))

print("\nMissing values:")
print(history.isnull().sum())

print("\nFirst five chatbot responses:")

for index, row in history.head(5).iterrows():
    print("Question:", row["User_Question"])
    print("Response:", row["Bot_Response"])
    print("-" * 40)
```

```python

history["Response_Length"] = history["Bot_Response"].str.len()

display(history[["User_Question", "Response_Length"]])

print("Average response length:",
      history["Response_Length"].mean())
```

```python

history.to_csv("../dataset/response_analysis.csv", index=False)

print("Response analysis saved successfully!")
```


## Observations

1. The chatbot generates responses using the Ollama language model.
2. Conversation history stores user questions and chatbot answers.
3. Timestamps help identify when conversations occurred.
4. Response lengths vary depending on the questions asked.
5. The chatbot may sometimes provide incomplete or incorrect answers.

## Strengths

## Limitations
